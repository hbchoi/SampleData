# 1. Gapminder 데이터 소개

Gapminder는 세계 여러 나라의 사회·경제적 변화를 이해할 수 있도록 다양한 국제 통계 지표를 제공하는 프로젝트입니다.

이번 실습에서는 여러 Gapminder 지표를 하나의 tabular dataset으로 결합한 데이터를 사용합니다.

- **대상:** 여러 국가
- **기간:** 2000, 2005, 2010, 2015, 2020
- **구조:** 한 국가의 한 시점이 하나의 row
- **내용:** 인구, 소득, 기대수명, 출산율, 아동사망률, CO₂ 배출, 빈곤, 교통사망, 인터넷 이용, 휴대전화 보급 등

따라서 한 가지 주제만 보는 것이 아니라, 서로 다른 변수들을 연결해서 다양한 질문을 만들어볼 수 있습니다.

예를 들면:

- 경제적으로 부유한 나라들은 기대수명도 높은 편일까?
- 출산율이 낮은 나라들은 어떤 특징을 가지고 있을까?
- 인터넷 이용이 빠르게 증가한 지역은 어디일까?
- CO₂ 배출이 많은 나라들은 소득 수준도 높은 편일까?
- 2000년과 2020년 사이에 가장 크게 변화한 국가는 어디일까?


# 2. 변수 설명

| 변수 | 의미 |
|---|---|
| `country_code` | 국가 코드 |
| `country` | 국가 이름 |
| `region` | Gapminder의 6개 지역 구분 |
| `year` | 관측 연도 |
| `population` | 총인구 |
| `income_per_capita` | 1인당 소득 |
| `life_expectancy` | 기대수명(년) |
| `fertility_rate` | 여성 1명당 평균 출생아 수 |
| `child_mortality` | 출생아 1,000명당 5세 미만 사망자 수 |
| `co2_per_capita` | 1인당 CO₂ 배출량 |
| `people_in_poverty` | 빈곤 상태에 있는 인구 수 |
| `traffic_deaths` | 인구 10만 명당 교통사망자 수 |
| `internet_users` | 인터넷 이용자 비율 |
| `cell_phones_total` | 휴대전화 가입/보유 총량 |

### Region 구분

`region`에는 다음과 같은 6개 지역이 포함됩니다.

- East Asia & Pacific
- Europe & Central Asia
- South Asia
- Middle East & North Africa
- Sub-Saharan Africa
- Americas

> **주의:** 일부 지표에는 결측치가 있을 수 있습니다.
