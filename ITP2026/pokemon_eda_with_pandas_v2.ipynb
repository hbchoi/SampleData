## 1. Load the Data

Pandas를 불러오고, Hugging Face에 공개된 CSV 파일을 웹에서 바로 읽어옵니다.

````python
import pandas as pd

url = "https://huggingface.co/datasets/Saif-M/Pokemon_Data/raw/main/Pokemon.csv"
pokemon = pd.read_csv(url)

pokemon.head()
````

## 3. Find Interesting Pokémon

먼저 조건에 맞는 Pokémon을 찾아봅니다.

````python
# Find Pikachu
pokemon[pokemon["name"] == "Pikachu"]
````

````python
# Pokémon with Attack >= 120
pokemon[pokemon["attack"] >= 120]
````

````python
# Top 10 Pokémon by Attack
pokemon.sort_values("attack", ascending=False).head(10)
````

## 4. Generation 1 vs. Generation 8

두 세대를 직접 **filtering**한 뒤 평균 능력치를 비교해봅니다.

````python
gen1 = pokemon[pokemon["generation"] == 1]
gen8 = pokemon[pokemon["generation"] == 8]

print("Generation 1:", len(gen1))
print("Generation 8:", len(gen8))
````

````python
stats = ["hp", "attack", "defense", "sp_attack", "sp_defense", "speed"]

gen1[stats].mean()
````

````python
gen8[stats].mean()
````

### Question

**Are newer Pokémon stronger?**

두 결과를 비교해보세요.

- 어느 세대의 평균 Attack이 더 높은가요?
- 어느 세대가 더 빠른가요?
- 모든 능력치에서 같은 세대가 더 높은가요?

````python
print("Generation 1 Total:", gen1["total"].mean())
print("Generation 8 Total:", gen8["total"].mean())
````

## 5. Fire vs. Electric

이번에는 Type을 기준으로 filtering해봅니다.

````python
fire = pokemon[pokemon["type1"] == "Fire"]
electric = pokemon[pokemon["type1"] == "Electric"]

print("Fire:", len(fire))
print("Electric:", len(electric))
````

````python
type_stats = ["hp", "attack", "defense", "speed"]

fire[type_stats].mean()
````

````python
electric[type_stats].mean()
````

### Question

**How are Fire and Electric Pokémon different?**

- Which type has higher average Attack?
- Which type has higher average Defense?
- Which type is faster?

## 6. Legendary vs. Normal

이번에는 `legendary` column을 사용해 두 그룹을 비교합니다.

````python
legendary = pokemon[pokemon["legendary"] == True]
normal = pokemon[pokemon["legendary"] == False]

print("Legendary:", len(legendary))
print("Normal:", len(normal))
````

````python
legendary[stats].mean()
````

````python
normal[stats].mean()
````

````python
print("Legendary Total:", legendary["total"].mean())
print("Normal Total:", normal["total"].mean())
````

## Wrap-up

오늘 사용한 Pandas의 핵심 패턴을 다시 확인해봅시다.

